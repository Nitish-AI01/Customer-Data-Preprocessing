```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age': [25, np.nan, 35, np.nan, np.nan, 40],
    'city': ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})

# Fill missing values
df['age'] = df['age'].fillna(df['age'].median())
df['city'] = df['city'].fillna(df['city'].mode()[0])

# Remove duplicates
df = df.drop_duplicates()

print("Cleaned Data:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder

# One-Hot Encoding for city
df = pd.get_dummies(df, columns=['city'])

# Label Encoding for gender
le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])

print("\nEncoded Data:")
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Select columns
num_cols = ['age', 'annual_income']

# Min-Max Scaling
minmax = MinMaxScaler()
df_minmax = df.copy()
df_minmax[num_cols] = minmax.fit_transform(df_minmax[num_cols])

print("\nMin-Max Scaled Data:")
print(df_minmax)

# Standard Scaling
standard = StandardScaler()
df_standard = df.copy()
df_standard[num_cols] = standard.fit_transform(df_standard[num_cols])

print("\nStandard Scaled Data:")
print(df_standard)
```

**When to use Min-Max Scaling and Standardisation**

**Min-Max Scaling:**

* Use when data is not normally distributed

* Needed for algorithms like KNN, Neural Networks

* Keeps values between 0 and 1

**Standardisation:**

* Use when data is normally distributed

* Works best for Linear Regression, Logistic Regression

* Centers data around mean = 0, std = 1
